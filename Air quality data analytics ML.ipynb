```python
feature_uk = uk_data[[ 'Ozone', 'Nitric oxide', 'Nitrogen dioxide',
       'Nitrogen oxides as nitrogen dioxide', 'Sulphur dioxide']]

X=np.asarray(feature_uk)
y=np.asarray(uk_data['Nitrogen dioxide'])

X[0:5]
                      
```

```python
#Divide the data for train and test part
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test= train_test_split(X,y, test_size=0.2, random_state=4)
X_train.shape
y_train.shape

X_test.shape
y_test.shape
```

```python
#MODELING SVM MACHINE LEARNING
```

```python
from sklearn import svm

 
classifier = svm.SVC(kernel='linear', gamma='auto', C=2)
classifier.fit(X_train, y_train)
classifier.predict(X_test)
y_predict = classifier.predict(X_test)
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test, y_predict))
```
